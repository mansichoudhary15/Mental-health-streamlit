#### Chart - 1 Age Distribution of Survey Respondents

````python

# Chart 1: Distribution of Respondents by Age

# Create a copy of Age as numeric
age_data = pd.to_numeric(df['Age'], errors='coerce')

# Keep only realistic age values
age_data = age_data[(age_data >= 18) & (age_data <= 100)]

# Plot the age distribution
plt.figure(figsize=(10, 6))
plt.hist(age_data, bins=15, edgecolor='black')

plt.xlabel('Age')
plt.ylabel('Number of Respondents')
plt.title('Distribution of Respondents by Age')
plt.grid(axis='y', alpha=0.3)

plt.show()
````

##### 1. Why did you pick the specific chart?

I selected a histogram because it is suitable for understanding the distribution of a numerical variable such as Age. It helps identify the most common age ranges and gives an overall view of the age distribution of survey respondents.

##### 2. What is/are the insight(s) found from the chart?

The chart shows that the respondents are concentrated mainly in the younger age groups, with the highest concentration around the late 20s to early 30s. The number of respondents decreases as age increases, with relatively fewer respondents in the older age groups. The distribution is therefore concentrated toward younger respondents.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. Understanding the age distribution helps in interpreting the survey results in the correct respondent context and can support age-aware workplace mental health initiatives. However, this chart only describes the survey respondents and should not be treated as a representation of the entire technology workforce.

#### Chart - 2 Gender Distribution

````python
# Chart 2: Gender Distribution

# Standardizing gender values for meaningful analysis
gender_data = df['Gender'].astype(str).str.strip().str.lower()

gender_data = gender_data.replace({
    'm': 'male',
    'male': 'male',
    'man': 'male',
    'cis male': 'male',
    'cis man': 'male',

    'f': 'female',
    'female': 'female',
    'woman': 'female',
    'cis female': 'female',
    'cis woman': 'female'
})

# Keep the main gender categories
gender_data = gender_data[gender_data.isin(['male', 'female'])]

# Plot
plt.figure(figsize=(8, 5))
sns.countplot(x=gender_data)

plt.xlabel('Gender')
plt.ylabel('Number of Respondents')
plt.title('Gender Distribution of Survey Respondents')

plt.show()
````

##### 1. Why did you pick the specific chart?

I selected a bar chart because Gender is a categorical variable. A bar chart makes it easy to compare the number of respondents across different gender categories and understand the overall composition of the survey sample.

##### 2. What is/are the insight(s) found from the chart?

The chart shows that male respondents form the majority of the survey sample, while female respondents represent a much smaller proportion. This indicates that the dataset has an uneven gender distribution, with substantially more male respondents than female respondents.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. Understanding the gender composition of the sample is important while interpreting other mental health findings. The results should be considered in the context of this uneven representation, and gender-based analysis can help identify whether workplace mental health experiences differ across the represented groups.

#### Chart - 6 Age Group vs Mental Health Treatment

````python
# Chart 6: Age Group vs Mental Health Treatment

# Create a working copy for this analysis
chart6_df = df.copy()

# Convert Age to numeric and keep realistic age values
chart6_df['Age'] = pd.to_numeric(chart6_df['Age'], errors='coerce')
chart6_df = chart6_df[
    (chart6_df['Age'] >= 18) &
    (chart6_df['Age'] <= 100)
].copy()

# Create age groups
chart6_df['Age_Group'] = pd.cut(
    chart6_df['Age'],
    bins=[17, 25, 35, 45, 55, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56+']
)

# Plot
plt.figure(figsize=(10, 6))

sns.countplot(
    data=chart6_df,
    x='Age_Group',
    hue='treatment'
)

plt.xlabel('Age Group')
plt.ylabel('Number of Respondents')
plt.title('Mental Health Treatment by Age Group')
plt.legend(title='Treatment')

plt.show()
````

##### 1. Why did you pick the specific chart?

I selected a grouped bar chart because Age Group and Treatment are categorical variables after grouping the numerical Age variable. This chart helps compare the number of respondents who reported seeking treatment across different age groups.

##### 2. What is/are the insight(s) found from the chart?

The 26–35 age group has the largest number of respondents, with 341 reporting treatment and 360 reporting no treatment. In the 36–45 age group, 153 respondents reported treatment compared with 124 who did not. The 46–55 and 56+ groups contain fewer respondents overall.

The treatment response is fairly balanced in the 18–25 group, while the 36–45 group has more respondents reporting treatment than no treatment. Overall, the chart shows that treatment responses vary across age groups, but the sample size differs considerably between groups.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. Understanding treatment patterns across age groups can help organizations consider whether mental health communication and support resources are reaching employees at different stages of their careers. However, the smaller number of respondents in older age groups means these comparisons should be interpreted carefully and should not be generalized to the entire workforce.
